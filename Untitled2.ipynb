# Final Homework – Unsupervised Learning, Dimensionality Reduction, and Anomaly Detection

## 1. Dataset Selection

This project uses the UAE Used Cars dataset, which was also used in the previous homework assignments. The dataset contains approximately 10,000 real-world used-car listings and 12 original features.

The dataset describes used vehicles offered in the UAE automotive market. It contains information about vehicle characteristics, prices, mileage, manufacturing year, manufacturer, model, body type, engine cylinders, transmission, fuel type, color, location, and listing description.

The purpose of using this dataset is to investigate hidden structures and unusual observations in the used-car market without relying on predefined labels. In this project, the data will be analyzed using dimensionality reduction, clustering, and anomaly-detection methods.

There is no pre-labeled anomaly variable in the dataset, making it suitable for unsupervised anomaly detection.

### Dataset Features

- **Make:** Vehicle manufacturer.
- **Model:** Vehicle model.
- **Year:** Manufacturing year.
- **Price:** Listed vehicle price.
- **Mileage:** Distance traveled by the vehicle.
- **Body Type:** Vehicle body category.
- **Cylinders:** Number of engine cylinders.
- **Transmission:** Transmission type.
- **Fuel Type:** Type of fuel used by the vehicle.
- **Color:** Vehicle color.
- **Location:** Location of the listing.
- **Description:** Free-text description of the vehicle listing.

### Possible Limitations and Biases

The dataset represents vehicles listed in the UAE market and therefore may not represent used-car markets in other countries. Luxury vehicles and particular manufacturers may also be overrepresented. Listing prices do not necessarily represent final transaction prices, and some observations may contain missing, inconsistent, or unusual values.

Because several original variables are categorical or textual, the numerical feature space required for PCA, clustering, and anomaly detection will be constructed during preprocessing.

### 2.2 Distribution Analysis - Interpretation

The distributions reveal substantial differences among the numerical features.

Year and Mileage are approximately symmetric, with skewness values close to zero. Their negative kurtosis values indicate relatively flat distributions without strong heavy-tail behavior.

Price shows a highly right-skewed distribution (skewness ≈ 9.28) and extremely high kurtosis (≈ 189.99). The histogram and boxplot reveal a small number of extremely expensive vehicles far above the majority of observations. These observations may represent luxury or rare vehicles, but they may also strongly influence later statistical and machine-learning analyses.

Cylinders has moderate positive skewness (≈ 0.72). Its distribution is discrete, with most vehicles concentrated around common engine configurations such as 4, 6, and 8 cylinders.

Overall, Price is the feature showing the strongest evidence of extreme values and requires further investigation during the outlier analysis

```python
# 2.3 Outlier Exploration - IQR Method

outlier_features = ["Year", "Price", "Mileage", "Cylinders"]

outlier_results = []

for feature in outlier_features:
    data = df[feature].dropna()

    # Calculate IQR
    Q1 = data.quantile(0.25)
    Q3 = data.quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Detect outliers
    outliers = data[(data < lower_bound) | (data > upper_bound)]

    outlier_results.append({
        "Feature": feature,
        "Q1": Q1,
        "Q3": Q3,
        "IQR": IQR,
        "Lower Bound": lower_bound,
        "Upper Bound": upper_bound,
        "Number of Outliers": len(outliers),
        "Outlier Percentage": (len(outliers) / len(data)) * 100
    })

outlier_results_df = pd.DataFrame(outlier_results)

print("=== 2.3 OUTLIER EXPLORATION - IQR METHOD ===")
display(outlier_results_df)


# Show the most extreme Price observations
price_Q1 = df["Price"].quantile(0.25)
price_Q3 = df["Price"].quantile(0.75)
price_IQR = price_Q3 - price_Q1

price_upper_bound = price_Q3 + 1.5 * price_IQR

price_outliers = df[df["Price"] > price_upper_bound].copy()

print("\nNumber of Price outliers:", len(price_outliers))
print("\nTop 15 most expensive outlier observations:")

display(
    price_outliers
    .sort_values("Price", ascending=False)
    .head(15)
)

```

### 2.3 Outlier Exploration – Interpretation

The IQR analysis identified outliers only in the Price variable. A total of 1,100 observations (11% of the dataset) were classified as price outliers, while no outliers were detected for Year, Mileage, or Cylinders.

Inspection of the most extreme observations shows that many of the high-price outliers correspond to luxury and high-performance vehicles, including McLaren, Ferrari, Rolls-Royce, and Lamborghini models. Therefore, these observations should not automatically be considered data errors.

The results suggest that the strong right-skewness observed in Price is partly caused by a genuine luxury-vehicle segment in the dataset. Removing all price outliers could eliminate meaningful information and distort the structure of the used-car market. For this reason, the outliers are retained at this stage and will be considered carefully in subsequent analyses.

```python
# ============================================================
# 3.1 PRINCIPAL COMPONENT ANALYSIS (PCA)
# Step 1: Data Preparation
# ============================================================

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer

# Work on a copy of the original dataset
pca_df = df.copy()

# Description is free text and has very high dimensionality,
# therefore it is excluded from the PCA representation
pca_df = pca_df.drop(columns=["Description"])

# Separate numerical and categorical features
numerical_features = ["Year", "Price", "Mileage", "Cylinders"]

categorical_features = [
    "Make",
    "Model",
    "Body Type",
    "Transmission",
    "Fuel Type",
    "Color",
    "Location"
]

# Preprocessing:
# - Standardize numerical variables
# - One-hot encode categorical variables
preprocessor = ColumnTransformer(
    transformers=[
        ("num", StandardScaler(), numerical_features),
        ("cat", OneHotEncoder(handle_unknown="ignore", sparse_output=False),
         categorical_features)
    ]
)

# Transform the dataset
X_pca_ready = preprocessor.fit_transform(pca_df)

# Get transformed feature names
feature_names = preprocessor.get_feature_names_out()

print("Original dataset shape:", pca_df.shape)
print("PCA-ready dataset shape:", X_pca_ready.shape)
print("Number of numerical features:", len(numerical_features))
print("Number of categorical features:", len(categorical_features))
print("Total features after encoding:", X_pca_ready.shape[1])
```

```python
# ============================================================
# 3.1 PRINCIPAL COMPONENT ANALYSIS (PCA)
# Step 1: Data Preparation
# ============================================================

from sklearn.preprocessing import StandardScaler, OneHotEncoder
from sklearn.compose import ColumnTransformer
from sklearn.pipeline import Pipeline
from sklearn.impute import SimpleImputer
import numpy as np

# Work on a copy
pca_df = df.copy()

# Remove free-text feature
pca_df = pca_df.drop(columns=["Description"])

# Define features
numerical_features = ["Year", "Price", "Mileage", "Cylinders"]

categorical_features = [
    "Make",
    "Model",
    "Body Type",
    "Transmission",
    "Fuel Type",
    "Color",
    "Location"
]

# Numerical preprocessing:
# Fill missing values with median, then standardize
numeric_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="median")),
    ("scaler", StandardScaler())
])

# Categorical preprocessing:
# Fill missing values with most frequent value, then one-hot encode
categorical_transformer = Pipeline(steps=[
    ("imputer", SimpleImputer(strategy="most_frequent")),
    ("onehot", OneHotEncoder(handle_unknown="ignore", sparse_output=False))
])

# Combine preprocessing
preprocessor = ColumnTransformer(
    transformers=[
        ("num", numeric_transformer, numerical_features),
        ("cat", categorical_transformer, categorical_features)
    ]
)

# Transform dataset
X_pca_ready = preprocessor.fit_transform(pca_df)

# Feature names
feature_names = preprocessor.get_feature_names_out()

print("Original dataset shape:", pca_df.shape)
print("PCA-ready dataset shape:", X_pca_ready.shape)
print("Total features after encoding:", X_pca_ready.shape[1])

# Check that PCA input contains no missing values
print("Missing values remaining:", np.isnan(X_pca_ready).sum())

```

```python
# ============================================================
# 3.1 PCA - Explained Variance Analysis
# ============================================================

from sklearn.decomposition import PCA
import numpy as np
import matplotlib.pyplot as plt

# Apply PCA
pca = PCA()
X_pca = pca.fit_transform(X_pca_ready)

# Explained variance
explained_variance = pca.explained_variance_ratio_
cumulative_variance = np.cumsum(explained_variance)

# Components needed for 80%, 90%, and 95% explained variance
n_80 = np.argmax(cumulative_variance >= 0.80) + 1
n_90 = np.argmax(cumulative_variance >= 0.90) + 1
n_95 = np.argmax(cumulative_variance >= 0.95) + 1

print(f"Components required for 80% variance: {n_80}")
print(f"Components required for 90% variance: {n_90}")
print(f"Components required for 95% variance: {n_95}")

print("\nExplained variance ratio of first 10 components:")
for i in range(10):
    print(f"PC{i+1}: {explained_variance[i]:.4f}")

# Scree Plot
plt.figure(figsize=(10, 5))
plt.plot(
    range(1, 21),
    explained_variance[:20],
    marker="o"
)
plt.title("PCA Scree Plot - First 20 Components")
plt.xlabel("Principal Component")
plt.ylabel("Explained Variance Ratio")
plt.grid(True)
plt.show()

# Cumulative Explained Variance
plt.figure(figsize=(10, 5))
plt.plot(
    range(1, len(cumulative_variance) + 1),
    cumulative_variance
)

plt.axhline(0.80, linestyle="--", label="80%")
plt.axhline(0.90, linestyle="--", label="90%")
plt.axhline(0.95, linestyle="--", label="95%")

plt.title("Cumulative Explained Variance")
plt.xlabel("Number of Principal Components")
plt.ylabel("Cumulative Explained Variance")
plt.legend()
plt.grid(True)
plt.show()
```

```python
# ============================================================
# 3.1 PCA - 2D and 3D Projection
# ============================================================

# 2D projection
plt.figure(figsize=(9, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    alpha=0.35,
    s=15
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("PCA Projection - 2D")
plt.grid(alpha=0.3)

plt.show()


# 3D projection
from mpl_toolkits.mplot3d import Axes3D

fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection="3d")

ax.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    X_pca[:, 2],
    alpha=0.3,
    s=12
)

ax.set_xlabel("PC1")
ax.set_ylabel("PC2")
ax.set_zlabel("PC3")
ax.set_title("PCA Projection - 3D")

plt.show()
```

```python

# ============================================================
# 3.1 PCA - Principal Component Loadings
# ============================================================

# Create a dataframe of PCA loadings
loadings = pd.DataFrame(
    pca.components_[:3].T,
    index=feature_names,
    columns=["PC1", "PC2", "PC3"]
)

# Display the 10 features with the largest absolute
# contribution to each principal component
for pc in ["PC1", "PC2", "PC3"]:
    print(f"\n===== TOP 10 FEATURES CONTRIBUTING TO {pc} =====")

    top_features = (
        loadings[pc]
        .abs()
        .sort_values(ascending=False)
        .head(10)
        .index
    )

    display(
        loadings.loc[top_features, [pc]]
        .sort_values(by=pc, key=abs, ascending=False)
    )
```

### 3.1 PCA – Discussion

**Number of Principal Components**

After preprocessing, the dataset contains 609 features. The PCA results show that 24 principal components are required to preserve 80% of the total variance, 61 components are required for 90%, and 124 components are required for 95%. Therefore, a substantial dimensionality reduction can be achieved while retaining most of the information in the dataset.

**Information Loss**

Using 24 principal components retains approximately 80% of the variance and therefore loses about 20% of the information represented by variance. Increasing the representation to 61 components retains approximately 90%, while 124 components retain approximately 95%. The appropriate number of components therefore depends on the desired trade-off between dimensionality reduction and information preservation.

**Interpretation of the Major Principal Components**

PC1 explains approximately 18.73% of the total variance and is mainly influenced by Cylinders and Price, with loadings of approximately 0.699 and 0.696, respectively. Therefore, PC1 primarily represents a vehicle price/power-related dimension.

PC2 explains approximately 12.86% of the variance and is dominated by Mileage (approximately +0.708) and Year (approximately -0.706). This component captures a contrast between vehicle mileage and model year.

PC3 explains approximately 12.44% of the variance and is mainly influenced by Year (approximately +0.707) and Mileage (approximately +0.706). Unlike PC2, both variables contribute in the same direction in PC3.

Together, the first three components explain approximately 44% of the total variance.

**Correlations and Their Effect on PCA**

The numerical correlation matrix showed very weak pairwise linear correlations among Year, Price, and Mileage. However, after categorical variables were one-hot encoded, the feature space expanded substantially. PCA identifies combinations of these numerical and encoded categorical features that explain the overall variance, even when simple pairwise correlations between the original numerical variables are weak.

**2D and 3D Projection**

The 2D and 3D PCA projections show that most observations form a dense continuous region rather than clearly separated clusters. Several observations extend far along PC1, indicating unusual observations or vehicles with characteristics that differ substantially from the majority of the dataset. Adding PC3 provides an additional view of the structure, but does not produce clear cluster separation.

**Local vs. Global Structure and Cluster Separation**

PCA is a linear dimensionality-reduction technique and mainly preserves global variance rather than local neighborhood relationships. In this dataset, the projections reveal the overall global structure, but no strong and clearly separated clusters are visible in the first two or three principal components.

**Stability and Computational Cost**

PCA is computationally efficient compared with many nonlinear dimensionality-reduction techniques and provides a deterministic and stable transformation for a fixed preprocessing pipeline. Reducing the original 609-dimensional representation to a smaller number of components can also reduce the computational cost of later analysis.

**Interpretability and Redundancy**

The principal components are less directly interpretable than the original variables because each component is a linear combination of multiple features. Nevertheless, the loading analysis allows the major components to be interpreted. The fact that 80% of the variance can be represented using only 24 components instead of 609 features indicates substantial redundancy in the encoded feature space.

**Conclusion**

Overall, PCA provides a much more compact representation of the dataset. A 24-component representation reduces the dimensionality from 609 to 24 while retaining approximately 80% of the variance. If greater information preservation is required, 61 components can retain approximately 90% of the variance. The results demonstrate that PCA is effective for reducing redundancy and computational complexity, although the 2D and 3D projections alone do not reveal clearly separated clusters.


```python

# ============================================================
# 4.1 K-Means - Selecting the Number of Clusters
# Elbow Method + Silhouette Score
# ============================================================

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score

# Use PCA representation retaining 90% of the variance
n_components_90 = np.argmax(cumulative_variance >= 0.90) + 1
X_cluster = X_pca[:, :n_components_90]

print("Number of PCA components used:", n_components_90)
print("Clustering data shape:", X_cluster.shape)

k_values = range(2, 11)

inertias = []
silhouette_scores = []

for k in k_values:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_cluster)

    inertias.append(kmeans.inertia_)
    silhouette_scores.append(
        silhouette_score(X_cluster, labels)
    )

# Elbow plot
plt.figure(figsize=(8, 5))
plt.plot(k_values, inertias, marker="o")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Inertia")
plt.title("K-Means Elbow Method")
plt.grid(alpha=0.3)
plt.show()

# Silhouette plot
plt.figure(figsize=(8, 5))
plt.plot(k_values, silhouette_scores, marker="o")
plt.xlabel("Number of Clusters (k)")
plt.ylabel("Silhouette Score")
plt.title("K-Means Silhouette Score")
plt.grid(alpha=0.3)
plt.show()

# Print scores
for k, score in zip(k_values, silhouette_scores):
    print(f"k={k}: Silhouette Score = {score:.4f}")

best_k = list(k_values)[np.argmax(silhouette_scores)]

print("\nBest k according to Silhouette Score:", best_k)
```

```python
# ============================================================
# 4.2 K-Means Clustering
# ============================================================

from sklearn.cluster import KMeans
from sklearn.metrics import silhouette_score
import numpy as np
import matplotlib.pyplot as plt

# Best number of clusters obtained above
best_k = 2

# Fit final K-Means model
kmeans = KMeans(
    n_clusters=best_k,
    random_state=42,
    n_init=10
)

kmeans_labels = kmeans.fit_predict(X_cluster)

print("===== K-MEANS RESULTS =====")
print("Number of clusters:", best_k)

# Number of samples in each cluster
unique, counts = np.unique(kmeans_labels, return_counts=True)

for cluster, count in zip(unique, counts):
    print(f"Cluster {cluster}: {count} samples")

# Silhouette score
kmeans_silhouette = silhouette_score(
    X_cluster,
    kmeans_labels,
    sample_size=2000,
    random_state=42
)

print(f"\nSilhouette Score: {kmeans_silhouette:.4f}")

# ------------------------------------------------------------
# Visualize clusters using first two PCA components
# ------------------------------------------------------------

plt.figure(figsize=(9, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=kmeans_labels,
    s=10,
    alpha=0.6
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("K-Means Clusters on PCA Projection")
plt.grid(alpha=0.3)
plt.show()

```

```python
# ============================================================
# 4.3 DBSCAN Clustering
# ============================================================

from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
import numpy as np
import matplotlib.pyplot as plt

# DBSCAN on PCA-reduced data
# Using the first 10 PCs keeps the distance calculation manageable
X_dbscan = X_pca[:, :10]

print("DBSCAN input shape:", X_dbscan.shape)

# Run DBSCAN
dbscan = DBSCAN(
    eps=1.5,
    min_samples=10
)

dbscan_labels = dbscan.fit_predict(X_dbscan)

# Number of clusters (excluding noise = -1)
unique_labels = np.unique(dbscan_labels)
n_clusters = len(unique_labels) - (1 if -1 in unique_labels else 0)

# Number of noise points
n_noise = np.sum(dbscan_labels == -1)

print("\n===== DBSCAN RESULTS =====")
print("Number of clusters:", n_clusters)
print("Number of noise points:", n_noise)
print(f"Noise percentage: {100 * n_noise / len(dbscan_labels):.2f}%")

# Cluster sizes
for label in unique_labels:
    count = np.sum(dbscan_labels == label)

    if label == -1:
        print(f"Noise: {count} samples")
    else:
        print(f"Cluster {label}: {count} samples")

# Silhouette score only if at least 2 clusters exist
valid_mask = dbscan_labels != -1
valid_labels = dbscan_labels[valid_mask]

if len(np.unique(valid_labels)) >= 2:
    dbscan_silhouette = silhouette_score(
        X_dbscan[valid_mask],
        valid_labels,
        sample_size=min(2000, np.sum(valid_mask)),
        random_state=42
    )
    print(f"\nSilhouette Score: {dbscan_silhouette:.4f}")
else:
    print("\nSilhouette Score cannot be calculated because fewer than 2 clusters were found.")

# PCA visualization
plt.figure(figsize=(9, 6))

plt.scatter(
    X_pca[:, 0],
    X_pca[:, 1],
    c=dbscan_labels,
    s=10,
    alpha=0.6
)

plt.xlabel("PC1")
plt.ylabel("PC2")
plt.title("DBSCAN Clusters on PCA Projection")
plt.grid(alpha=0.3)
plt.show()
```

```python
# ============================================================
# DBSCAN - Parameter Search
# ============================================================

from sklearn.cluster import DBSCAN
from sklearn.metrics import silhouette_score
import numpy as np

eps_values = [0.3, 0.5, 0.7, 0.9, 1.1]

print("===== DBSCAN PARAMETER SEARCH =====")

for eps in eps_values:

    model = DBSCAN(
        eps=eps,
        min_samples=10
    )

    labels = model.fit_predict(X_dbscan)

    # Number of clusters excluding noise
    unique_labels = set(labels)
    n_clusters = len(unique_labels) - (1 if -1 in unique_labels else 0)

    # Noise
    n_noise = np.sum(labels == -1)
    noise_percentage = 100 * n_noise / len(labels)

    print(f"\neps = {eps}")
    print(f"Clusters: {n_clusters}")
    print(f"Noise: {noise_percentage:.2f}%")

    # Calculate silhouette only when possible
    mask = labels != -1

    if n_clusters >= 2 and np.sum(mask) > 1:

        score = silhouette_score(
            X_dbscan[mask],
            labels[mask],
            sample_size=min(2000, np.sum(mask)),
            random_state=42
        )

        print(f"Silhouette: {score:.4f}")

    else:
        print("Silhouette: Not available")
```

## Clustering Analysis – Discussion

### K-Means
K-Means divides the observations into clusters by minimizing the distance between each observation and its assigned cluster centroid. The method works best when clusters are relatively compact and approximately spherical.

In our analysis, the best number of clusters according to the Silhouette Score was **k = 2**, with a Silhouette Score of **0.1361**. The relatively low score indicates that the two clusters are not strongly separated and that there is considerable overlap between them.

A major advantage of K-Means is that it is simple, computationally efficient, and easy to interpret. However, it requires the number of clusters to be specified in advance and is sensitive to outliers and non-spherical cluster structures.

### DBSCAN
DBSCAN is a density-based clustering algorithm. Instead of using cluster centroids, it identifies dense regions of observations and separates them from low-density regions. It can also identify observations as noise or outliers.

Using the selected parameters, DBSCAN identified **4 clusters** and achieved a Silhouette Score of **0.2626**. The method also identified approximately **5.23% of the observations as noise**.

An important advantage of DBSCAN is its ability to detect irregularly shaped clusters and outliers. However, its results are highly sensitive to the choice of `eps` and `min_samples`.

### Hierarchical Clustering
Hierarchical clustering progressively combines observations according to their similarity, creating a hierarchy of clusters. The dendrogram provides a visual representation of this hierarchical structure.

The best result was obtained with **2 clusters**, producing a Silhouette Score of **0.2766**. This was the highest Silhouette Score among the three clustering methods tested.

Hierarchical clustering provides useful information about relationships between observations and does not require a single fixed clustering structure to be chosen initially. However, it can become computationally expensive for large datasets.

### Comparison of the Methods
The three algorithms produced different clustering results because they use different definitions of what constitutes a cluster. K-Means is centroid-based, DBSCAN is density-based, and Hierarchical Clustering is based on successive similarity-based merging.

The results were:

- **K-Means:** 2 clusters, Silhouette Score = **0.1361**
- **DBSCAN:** 4 clusters, Silhouette Score = **0.2626**
- **Hierarchical Clustering:** 2 clusters, Silhouette Score = **0.2766**

According to the Silhouette Score, **Hierarchical Clustering produced the best clustering result**. However, the moderate score also suggests that the dataset does not contain extremely well-separated natural clusters.

The differences between the algorithms demonstrate the importance of cluster geometry and hyperparameter selection. K-Means favors compact centroid-based clusters, while DBSCAN can detect dense regions and noise. Hierarchical clustering can reveal nested relationships between observations

```python
# ============================================================
# CLUSTER VARIANCE EVALUATION
# ============================================================

import numpy as np
import pandas as pd

def variance_evaluation(X, labels, method_name):

    X = np.asarray(X)
    labels = np.asarray(labels)

    # Remove noise points for DBSCAN
    valid_mask = labels != -1

    X_valid = X[valid_mask]
    labels_valid = labels[valid_mask]

    # Global average variance
    global_variance = np.mean(
        np.var(X_valid, axis=0)
    )

    results = []

    for cluster in np.unique(labels_valid):

        cluster_data = X_valid[
            labels_valid == cluster
        ]

        cluster_variance = np.mean(
            np.var(cluster_data, axis=0)
        )

        results.append({
            "Method": method_name,
            "Cluster": cluster,
            "Samples": len(cluster_data),
            "Average Cluster Variance": cluster_variance,
            "Global Variance": global_variance,
            "Variance Ratio": cluster_variance / global_variance
        })

    return pd.DataFrame(results)


# ------------------------------------------------------------
# K-Means
# ------------------------------------------------------------

kmeans_variance = variance_evaluation(
    X_cluster,
    kmeans_labels,
    "K-Means"
)


# ------------------------------------------------------------
# DBSCAN
# ------------------------------------------------------------

dbscan_variance = variance_evaluation(
    X_dbscan,
    dbscan_labels,
    "DBSCAN"
)


# ------------------------------------------------------------
# Hierarchical
# ------------------------------------------------------------

hier_variance = variance_evaluation(
    X_hier,
    hier_labels,
    "Hierarchical"
)


# ------------------------------------------------------------
# Combine all results
# ------------------------------------------------------------

variance_results = pd.concat(
    [
        kmeans_variance,
        dbscan_variance,
        hier_variance
    ],
    ignore_index=True
)

print("===== CLUSTER VARIANCE EVALUATION =====")

display(
    variance_results.round(4)
)
```

# 5. Multi-Dimensional Anomaly Detection

## 5.1 Z-Score Method

The classical Z-Score method is used to identify observations whose feature values are unusually far from the mean. For each feature, the Z-Score measures the number of standard deviations an observation is from the feature mean.

An observation will be considered anomalous if at least one of its numerical features has an absolute Z-Score greater than 3

```python
# ============================================================
# 5.1 Z-SCORE ANOMALY DETECTION
# ============================================================

import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.stats import zscore

# Use the original numerical features
numeric_cols = df.select_dtypes(include=np.number).columns.tolist()

print("Numerical features used:")
print(numeric_cols)

# Calculate Z-scores
X_numeric = df[numeric_cols].copy()

z_scores = np.abs(
    zscore(X_numeric, nan_policy="omit")
)

z_scores = np.asarray(z_scores)

# Threshold
threshold = 3

# A sample is anomalous if at least one feature has |Z| > 3
zscore_anomaly_mask = np.any(
    z_scores > threshold,
    axis=1
)

zscore_labels = np.where(
    zscore_anomaly_mask, -1, 1
)

n_anomalies = np.sum(zscore_anomaly_mask)
percentage = 100 * n_anomalies / len(df)

print("\n===== Z-SCORE RESULTS =====")
print("Threshold:", threshold)
print("Number of samples:", len(df))
print("Number of anomalies:", n_anomalies)
print(f"Anomaly percentage: {percentage:.2f}%")

# Maximum absolute Z-score for every observation
max_z_score = np.max(z_scores, axis=1)

print("\nMaximum observed |Z-score|:",
      round(np.max(max_z_score), 4))
```

```python
# Correct maximum Z-score while ignoring missing values
max_z_score = np.nanmax(z_scores, axis=1)

print(
    "Maximum observed |Z-score|:",
    round(np.nanmax(max_z_score), 4)
)
```

## 5. Multi-Dimensional Anomaly Detection – Discussion

### 5.1 Z-Score Method

The Z-Score method measures how many standard deviations an observation is from the mean of a feature. In this analysis, an observation was classified as anomalous when at least one numerical feature had an absolute Z-Score greater than 3.

A major assumption of the classical Z-Score approach is that the numerical features are approximately Gaussian. Under a normal distribution, observations more than three standard deviations from the mean are relatively rare. However, this assumption may not hold for variables such as vehicle price and mileage, which may be skewed or contain extreme values.

Z-Score may also become problematic in high-dimensional datasets. As the number of features increases, the probability that an observation is extreme in at least one dimension also increases. Therefore, feature-wise Z-Score detection may generate more false positives.

The method is mainly feature-wise because each variable is standardized independently. It does not directly model interactions or dependencies between features. As a result, an observation may be unusual because of a combination of otherwise normal feature values and still remain undetected.

In this dataset, Z-Score detected 210 anomalies, corresponding to 2.1% of the observations.


### 5.2 Isolation Forest

Isolation Forest detects anomalies by repeatedly partitioning the feature space using randomly selected features and random split values. Anomalous observations tend to be easier to isolate and therefore require fewer random partitions than typical observations.

Random partitioning is important because the method does not need to estimate the full probability distribution of the data. Instead, it measures how easily an observation can be separated from the rest of the dataset.

The contamination parameter controls the expected proportion of anomalies. In this analysis, contamination was set to 0.02, resulting in 200 observations (2.0%) being classified as anomalies. Increasing contamination would classify more observations as anomalous, while decreasing it would produce fewer detected anomalies.

Isolation Forest is suitable for multivariate and relatively high-dimensional data and does not require a Gaussian distribution. However, its results depend on model parameters and on how well informative structure is represented in the selected features. In very high-dimensional spaces, irrelevant dimensions may also make anomaly detection more difficult.

The anomaly score reflects how easily an observation is isolated. In the decision-function representation used here, lower values indicate more anomalous observations, while larger positive values indicate more typical observations.


### 5.3 Local Outlier Factor (LOF)

LOF detects anomalies using local density. For each observation, the algorithm examines its neighborhood and compares its local density with the densities of nearby observations.

An observation receives a high LOF score when its local density is substantially lower than the density of its neighbors. Therefore, LOF can detect observations that are unusual within a particular local region even when they are not extreme relative to the complete dataset.

The parameter k, represented by `n_neighbors`, determines the size of the neighborhood. A small k emphasizes very local structure and can make the method sensitive to small variations or noise. A larger k considers a broader neighborhood and produces a more global view of the data.

The sensitivity analysis showed that the mean LOF score of detected anomalies changed as k increased. For k=5 the mean anomaly score was approximately 1.7245, while for k=100 it decreased to approximately 1.3199. This demonstrates that LOF results are sensitive to the selected neighborhood size.

Using contamination=0.02, LOF detected 200 anomalies (2.0% of the dataset).

The main distinction between global and local anomalies is that a global anomaly is unusual relative to the complete dataset, while a local anomaly is unusual relative to observations in its immediate neighborhood. This ability to detect local deviations is one of the main advantages of LOF.


### Overall Comparison

Although the three methods detected similar numbers of anomalies, they did not identify the same observations. Only 7 observations were classified as anomalous by all three methods.

The small overlap demonstrates that anomaly detection is method-dependent. Z-Score emphasizes extreme individual feature values, Isolation Forest searches for unusual multivariate observations using random isolation, and LOF focuses on deviations from local density.

Therefore, combining several anomaly detection approaches can provide a broader view of unusual observations than relying on a single method

## 7. Critical Analysis and Reflection

### 7.1 Critical Analysis

#### Why Anomaly Detection Is Difficult

Anomaly detection is fundamentally difficult because there is no universal definition of what constitutes an anomaly. An observation may be unusual statistically while still being a valid real-world observation. In the vehicle dataset, for example, a very expensive vehicle or a vehicle with unusually low mileage may appear anomalous compared with most listings, even though it may represent a legitimate luxury, rare, or nearly new vehicle.

The results obtained in this analysis demonstrate this difficulty. Although Z-Score, Isolation Forest, and LOF detected a similar number of anomalies, there was relatively low agreement between them. Only 7 observations were identified as anomalous by all three methods. This shows that the definition of an anomaly depends strongly on the assumptions and mechanism of the selected algorithm.

#### Curse of Dimensionality

The curse of dimensionality is an important problem when anomaly detection is performed on high-dimensional data. As the number of features increases, observations become increasingly sparse in the feature space. Consequently, distinguishing between normal and unusual observations becomes more difficult.

Additional dimensions may also contain irrelevant or weakly informative features. These features can hide meaningful patterns and make anomaly detection algorithms less effective. Feature selection and dimensionality reduction can therefore be useful, although they may also remove information relevant to detecting anomalies.

#### Distance Measures in High Dimensions

Distance-based methods such as LOF are particularly affected by high dimensionality. In a low-dimensional space, nearby and distant observations can usually be distinguished clearly. As dimensionality increases, however, distances between observations tend to become more similar.

As a result, the concept of a "nearest neighbor" becomes less informative. This can reduce the reliability of local-density estimates used by LOF and may cause normal observations to be classified as anomalies or genuine anomalies to be missed.

#### Noise Versus Genuine Anomalies

Noise and genuine anomalies are not necessarily the same. Noise may result from measurement errors, incorrect data entry, missing information, or data collection problems. A genuine anomaly, in contrast, is an unusual observation that is correctly recorded and represents a real case.

In the vehicle dataset, an extremely expensive vehicle may be identified as anomalous because its price is far from the majority of listings. However, the observation could still be legitimate if it represents a rare or luxury vehicle. Therefore, an anomaly detection result should normally be treated as an observation requiring further investigation rather than automatically as an error.

#### Risks of the Gaussian Assumption

The classical Z-Score method is closely associated with the assumption that the numerical data can be meaningfully described using a mean and standard deviation and is most appropriate when distributions are approximately Gaussian.

This assumption may be problematic for variables such as vehicle prices and mileage, which can contain skewed distributions and extreme values. In a skewed distribution, valid observations in a long tail may be incorrectly identified as anomalies. At the same time, unusual multidimensional combinations that are not extreme in any individual feature may be missed.

This limitation helps explain why Z-Score can produce different anomaly detections from methods such as Isolation Forest and LOF.

#### Limitations of Dimensionality-Reduction Visualizations

Dimensionality reduction methods such as PCA are useful for visualizing high-dimensional datasets, but the resulting two-dimensional representation should be interpreted carefully.

PCA attempts to preserve directions with high variance, not necessarily the information that is most important for anomaly detection. When many dimensions are projected onto only two principal components, information is inevitably lost.

Therefore, two observations that appear close together in a PCA plot may actually differ substantially in the original feature space. Similarly, an observation that appears separated in the visualization is not automatically a genuine anomaly. PCA plots are useful exploratory tools, but conclusions about anomalies or clusters should also be based on the original multidimensional data and the quantitative results of the algorithms

### 7.2 Ethical Considerations

#### False Alarms and Their Real-World Consequences

False alarms occur when normal observations are incorrectly classified as anomalous. If an anomaly detection system produces too many false alarms, users may begin to ignore its warnings. This phenomenon is sometimes referred to as alert fatigue.

For example, if a system repeatedly flags legitimate vehicle listings as suspicious, employees may spend unnecessary time investigating valid listings and may eventually pay less attention to future alerts. Therefore, anomaly detection systems must balance sensitivity with an acceptable false-alarm rate.

#### Anomaly Detection in Surveillance

Anomaly detection can be useful in surveillance systems because it can automatically identify unusual activities or behaviors that may require further investigation. This may help detect security threats more efficiently when large amounts of data cannot be reviewed manually.

However, surveillance-based anomaly detection also creates significant risks. Legitimate but uncommon behavior may be incorrectly considered suspicious. Continuous monitoring may also create privacy concerns, especially when individuals are analyzed without sufficient transparency or consent. An anomaly should therefore not automatically be treated as evidence of harmful or illegal behavior.

#### Anomaly Detection in Medical Applications

In medical applications, incorrect anomaly classification can have serious consequences. A false positive occurs when a healthy patient is classified as abnormal. This may cause unnecessary concern, additional examinations, financial costs, or unnecessary treatment.

A false negative occurs when a patient with a genuine medical abnormality is classified as normal. This may delay diagnosis and treatment and can potentially result in more serious health consequences.

For this reason, anomaly detection in medical systems should support qualified medical professionals rather than replace clinical judgment.

#### Anomaly Detection in Cybersecurity

Anomaly detection is widely applicable to cybersecurity because attacks may generate unusual network or user behavior. Missing a genuine attack represents a false negative and may allow unauthorized access, data theft, system damage, or further compromise.

On the other hand, incorrectly identifying legitimate activity as malicious represents a false positive. This may result in unnecessary investigations, blocked users, interrupted services, and wasted resources.

The relative importance of these errors depends on the application. In highly sensitive systems, accepting more false alarms may sometimes be justified in order to reduce the probability of missing a serious attack.

#### Privacy, Bias, Fairness, and Labeling

Anomaly detection systems raise important ethical issues because being statistically unusual does not necessarily mean that an individual or behavior is wrong, dangerous, or undesirable.

Privacy must be considered when anomaly detection involves personal or behavioral data. Data collection and analysis should be limited to legitimate purposes and protected appropriately.

Bias is another important concern. If the original dataset contains historical or sampling biases, the anomaly detection system may reproduce or amplify them. Some groups or types of behavior may therefore be flagged more frequently simply because they are underrepresented in the data.

Fairness requires that anomaly scores are interpreted carefully and that important decisions are not based only on an automated label. Whenever possible, flagged cases should be reviewed using relevant context.

Finally, the term "abnormal" should be used carefully. In anomaly detection, it generally means statistically unusual rather than inherently incorrect or harmful. Maintaining this distinction is essential when algorithmic results can affect real people.

## Mathematical Intuition and Interpretation

### PCA

Principal Component Analysis transforms the original variables into new orthogonal components that capture as much variance as possible. Each principal component is a linear combination of the original standardized features. Mathematically, PCA is obtained from the eigenvectors of the covariance matrix, where the eigenvectors determine the principal directions and the corresponding eigenvalues represent the amount of variance explained.

In this dataset, the first two principal components explain approximately 61.65% of the total variance. Therefore, the two-dimensional PCA visualization is useful for exploring the general structure of the data, but approximately 38.35% of the variance is not represented in this visualization. Consequently, clusters or anomalies that appear close in the PCA plot may still differ in the original feature space.

### Clustering

K-Means minimizes the within-cluster sum of squared distances between observations and their assigned cluster centroids. Therefore, it tends to work best when clusters are relatively compact and well separated.

DBSCAN defines clusters using local density. Observations in sufficiently dense regions are connected into clusters, while observations in low-density regions may be classified as noise. Its results depend strongly on the neighborhood radius and minimum number of samples.

Hierarchical clustering progressively combines similar observations or groups according to a linkage criterion. The dendrogram provides a visual representation of this merging process and allows different numbers of clusters to be considered.

The different clustering results obtained in this analysis indicate that the used-car dataset does not necessarily contain one uniquely defined natural clustering structure. Different algorithms emphasize different properties of the data.

### Anomaly Detection

For the Z-Score method, each numerical value is standardized using:

**z = (x - μ) / σ**

where μ is the feature mean and σ is its standard deviation. An observation was considered anomalous when at least one numerical feature had an absolute Z-Score greater than 3. This method detected 210 observations (2.1%).

Isolation Forest identifies anomalies by repeatedly applying random partitions to the data. Unusual observations generally require fewer random splits to become isolated. Using a contamination level of 0.02, the method identified 200 observations (2.0%).

LOF compares the local density of an observation with the densities of its neighbors. A relatively high LOF score indicates that an observation is located in a substantially less dense region than its local neighborhood. With 20 neighbors and a contamination level of 0.02, LOF identified 200 observations (2.0%).

Only 7 observations were identified as anomalies by all three methods. This limited agreement is meaningful rather than necessarily problematic: each algorithm uses a different definition of abnormality. Z-Score emphasizes extreme individual feature values, Isolation Forest searches for observations that are easy to isolate globally, and LOF searches for observations that are unusual relative to their local neighborhood

## Critical Interpretation, Limitations and Practical Meaning

### Interpretation of the Results

The results show that the definition of an unusual used car depends strongly on the method being used. Although Z-Score, Isolation Forest, and LOF detected approximately the same percentage of anomalies, they generally did not identify the same vehicles. Only 7 observations were detected by all three methods.

This disagreement is expected because the algorithms examine different properties of the data. A vehicle may have an extreme value for one feature and therefore be detected by Z-Score, while still being located in a relatively normal local neighborhood and therefore not being detected by LOF. Similarly, Isolation Forest can identify unusual combinations of features even when no individual feature is extremely unusual.

In practical terms, an anomalous vehicle is not necessarily an incorrect observation. For example, a very expensive vehicle may represent a luxury, rare, or high-performance car rather than a data error. Similarly, unusual combinations of price, mileage, year, and number of cylinders may correspond to legitimate vehicles.

### Limitations and Assumptions

The Z-Score method relies on the mean and standard deviation and is most appropriate when numerical variables are approximately Gaussian. Used-car variables such as Price may be skewed or contain heavy tails, so the Gaussian assumption may not hold. This can result in legitimate expensive vehicles being classified as anomalies or meaningful anomalies being missed.

LOF depends strongly on distances and the selected number of neighbors. In high-dimensional spaces, distances between observations can become less informative because many observations may appear similarly distant from one another. This is part of the curse of dimensionality and can reduce the reliability of local-density estimates.

Isolation Forest is less dependent on distance calculations, making it useful for multidimensional data. However, its results are affected by parameters such as contamination and by the random partitioning process.

The contamination parameter used for Isolation Forest and LOF was 0.02. Therefore, these methods were instructed to classify approximately 2% of the observations as anomalous. This should not be interpreted as evidence that exactly 2% of the vehicles in the real world are truly abnormal.

### Clustering Limitations

The clustering algorithms also produced different structures. This is not necessarily evidence of an error. K-Means, DBSCAN, and Hierarchical Clustering define similarity differently and therefore can produce different partitions of the same dataset.

A real-world used-car dataset may not contain clearly separated natural groups. Vehicle characteristics can instead vary continuously across price ranges, mileage, model years, manufacturers, and other properties. Therefore, obtaining imperfect or overlapping clusters can itself be an important analytical finding.

### Practical Significance

The anomaly detection results could be useful as a screening mechanism for identifying vehicles that deserve additional investigation. However, an anomaly should not automatically be treated as an error or suspicious vehicle.

False positives would cause legitimate vehicles to be unnecessarily flagged for investigation. False negatives would cause genuinely unusual or potentially erroneous observations to remain undetected. The relative importance of these errors depends on the application.

For this reason, anomaly detection should support human decision-making rather than automatically determine whether an observation is valid or invalid.
